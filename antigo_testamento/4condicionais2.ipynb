```python
# rank

rank_nescessario_fase1 = 5
rank_nescessario_fase2 = 10
rank_jogador = int(input("qual seu rank? "))

if rank_nescessario_fase1 <= rank_jogador < rank_nescessario_fase2:
  print("voce esta na fase 1")
elif rank_jogador >= rank_nescessario_fase2:
  print("voce esta na fase 2")
else:
  print("voce nao entrou para a competicao")
```

```python
saldo = 500

print("===Banco Colab===")
print(f"seu saldo é {saldo}")

saque = int(input("quanto deseja sacar?"))

if saque <= saldo:
  saldo = saldo - saque
  print("saque realizado!")
  print(f"saldo restante: R$ {saldo}")
else:
  print("saldo insuficiente!")
  print(f"se saldo continua sendo R$ {saldo}")
```
