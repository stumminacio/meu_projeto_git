```python
#1) Crie um programa que, dada uma string com uma frase informada pelo usuário (incluindo espaços em branco), conte:
#quantos espaços em branco existem na frase.
#quantas vezes aparecem as vogais a, e, i, o, u

#2)vertical e escada

#3) Faça um programa que recebe o nome completo do usuário e imprime:
#Só o primeiro nome;
#Sobrenome completo;
#Apenas o último nome.
```

```python
# questão 1

frase = input("Digite uma frase: ")

espacos = frase.count(' ')
virgulas = frase.count(',')
pontos = frase.count('.')
character = len(frase)

vogais = "aeiouAEIOUáéíóúÁÉÍÓÚàèìòùÀÈÌÒÙãõÃÕâêîôûÂÊÎÔÛ"
contador_vogais = 0
for char in frase:
    if char in vogais:
        contador_vogais += 1

print(f"número de espaços em branco = {espacos}")
print(f"número de vogáis na frase = {contador_vogais}")
print(f"número de vírgulas na frase = {virgulas}")
print(f"número de pontos na frase = {pontos}")
print(f"número de character na frase = {character}")
```

```python
# quetão 2

nome = "Ctrl + Play"

#vertical

print("---Vertical---")
for char in nome:
    print(char.upper())

print("\n" + "="*15 + "\n")

#escadinha
print("---Escada---")
for i in range(1, len(nome) + 1):
    print(nome[0:i].upper())
```

```python
# questão 3

nome_completo = input("Digite seu nome completo: ")

# Divide o nome uma única vez e guarda na variável 'partes'
partes = nome_completo.split()

# Verifica se o usuário digitou pelo menos uma palavra
if len(partes) > 0:
    print(f"Seu primeiro nome é: {partes[0]}")

    # Só tenta mostrar o sobrenome se houver mais de uma palavra
    if len(partes) > 1:
        print(f"Seu sobrenome (segundo nome) é: {partes[1]}")
    else:
        print("Seu sobrenome: Você digitou apenas um nome.")

    print(f"Seu último nome é: {partes[-1]}")
else:
    print("Você não digitou nenhum nome!")
```
