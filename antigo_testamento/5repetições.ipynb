# Aula

```python
# While = Enquanto
# Contagem crescente

contador = 1

while contador <= 10:
  print(contador)
  contador = contador + 1
contador = contador - 1
```

```python
# Contagem Regressiva

contador = 10

while contador > 0:
  print(contador)
  contador = contador - 1
```

```python
# porta de entrada

senha = ""

while senha != "abcd":
  senha = input("Digite sua senha: ")
  if senha != "abcd":
    print("senha incorreta!")
  elif senha == "abcd":
    print("senha correta, você entrou na conta!")
```

```python
# For = Quantidade especifica de ações
# A variável numero, se não definida, sempre começa em 0

for numero in range(5):
  print(numero)
```

```python
# Determinando uma sequência

for numero in range(0, 11, 2):
  print(numero)
```

```python
# Mostrando Letras de uma String

palavra = "python"

for letra in palavra:
  print(letra)
```

```python
# Percorrendo Listas

pessoas = ["inacio", "João", "Ricardo", "Fulano"]

for pessoa in pessoas:
  print(pessoa)
```

```python
# Usando o commando  "break", ele quebra a repetição

for numero in range(10):
  print(numero)

  if numero == 5:
    break
```

```python
# Exercício 1
# Mostre os numeros de 1 até 10 utulizando while

numero = 1

while numero <= 10:
  print(numero)
  numero = numero + 1
numero = numero - 1
```

```python
# Exercício 2
# Mostre os numeros de 1 até 10 utulizando for

for numero in range(11):
  print(numero)
```

```python
# Exercício 3
# Peça ao usuario uma palavra.
# mostre cada letra dela em linha distintas.

palavra = input("digite sua palavra: ")

for letra in palavra:
  print(letra)
```

```python
# Exercício 4
# Criar um campo de senha e, enquanto essa senha for incorreta,
# exiba "senha incorreta",
# se não, exiba "senha correta"

senha = ""

while senha != "senha123":
  senha = input("digite sua senha: ")
  if senha !=  "senha123":
    print("senha incorreta!, tente novamente: ")
  elif senha == "senha123":
    print("senha correta, você acessou sua conta!")
    break
```
