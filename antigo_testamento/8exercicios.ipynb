```python
numero = float(input("qual é o seu numero? "))

if numero / 25 <= 1 and numero / 25 >= 0:
  print(f"intervalo (0, 25]")
elif numero / 50 <= 1 and numero / 50 > 0:
  print(f"intervalo (25, 50]")
elif numero / 75 <= 1 and numero / 75 > 0:
  print(f"intervalo (50, 75]")
elif numero / 100 <= 1 and numero / 100 > 0:
  print(f"intervalo (75, 100]")
else:
  print("fora de intervalo")
```

```python
hora_1 = int(input("qual hora comecou o jogo?: "))
hora_2 = int(input("qual hora terminou o jogo?: "))

if hora_1 > 24 or\
   hora_1 < 1 or\
   hora_2 > 24 or\
   hora_2 < 1 :
   print("está errado, alguma hora foi digitada incorretamente, verifique...")

resultado_1 = hora_1
resultado_2 = 0
resultado_atual = resultado_1
resultado_final = 0

while resultado_atual != hora_2:
  resultado_atual = resultado_atual + 1
  resultado_final = resultado_final + 1
  if resultado_atual == 24:
    resultado_atual = resultado_2

if resultado_final == 0:
  resultado_final = 24

print(resultado_final)
```
