```python
'''

=== TEMA PARA CASA ===

Crie um programa que:

1.
Cadastre o herói

Peça:

Nome
2.
Crie um inventário

Comece com:

Espada
Poção
3.
Pergunte um novo item

Adicione o item à lista.

4.
Crie uma função de ataque

A função deve receber:

ataque
defesa

e calcular o dano.

5.
Crie um inimigo

O inimigo começa com:

100 de vida
6.

Enquanto o inimigo tiver vida:

Ataque!

O jogador deverá informar o dano causado.

O inimigo perde vida.

7.

Se a vida chegar a 0:

Inimigo derrotado!
'''
```

```python
heroi = input("Qual será o nome do herói? ")

inv = ["Espada", "Poção"]

print(f"Seu inventario está assim: {inv}")
opcao = input("Deseja adicionar um novo item? [s/n] ")

if opcao == "s":
  novo_item = input("Qual será o novo item? ")
  inv.append(novo_item)

def calcular_dano(ataque, defesa):
  dano = ataque - defesa
  if dano <= 0:
    print("você não deu dano, ataque baixo demais!")
  else:
    print(f"você deu {dano} de dano")
  return dano

vida_inimigo = 100

print("voce encontrou um inimigo! Ele possui 100 de vida, informe seu dano e a defesa de seu inimigo para causar dano :")

dano = calcular_dano(int(input("quanto de ataque você tem? ")),
              int(input("quanto de defesa o inimigo tem? ")))


while vida_inimigo >= 0:
  vida_inimigo = vida_inimigo - dano
  print(f"o inimigo ficou com {vida_inimigo} de vida, continue atacando! ")
  dano = calcular_dano(int(input("quanto de ataque você tem? ")),
              int(input("quanto de defesa o inimigo tem? ")))

print("voce derrotou o inimigo!!!")
```
