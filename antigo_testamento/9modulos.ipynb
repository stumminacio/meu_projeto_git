```python
'''
Modulos com a professora Débora

import {modulo} = importa algum modulo de fora do codigo

no modulo {math},
  math.sqrt = raiz quadrada, sqrt = square root
  math.ceil = arrendonda a parte decimal para cima, ceil = teto?
  math.trunc = mostra o valor bruto do numero

no modulo {random},
  random.randint = numero inteiro aleatorio entre algum numero dado pelo usuario
  random.choice = escolhe alguma opção predefinida pelo usuario

no modulo {time},
  time.sleep = aguarda o tempo(segundos) que o usuario colocar

tambem tem o datetime,
  ele serve para verificar a hora, dia, ano, mês

'''
```

```python
import math

num = float(input("digite um numero decimal: "))

print("Raiz quadrada: ", math.sqrt(num))
print("Arrendodado para cima: ", math.ceil(num))
print("Parte inteira", math.trunc(num))
```

```python
import random

numero_sorteado = random.randint(1, 10)
print(f"o numero sorteado foi: {numero_sorteado}")

aluno = random.choice(["Ana", "Carlos", "Beatriz", "João"])
print(f"o aluno escolhido foi: {aluno}")
```

```python
dado = random.randint(1, 6)
print(f"Você tirou {dado}")
```

```python
import time

print("contagem regressiva...")
print("3")
time.sleep(1)

print("2")
time.sleep(1)

print("1")
time.sleep(1)

print("Decolagem!")
```

```python
tempo = int(input("qual sera o tempo de contagem (regressiva): "))
tempo_atual = tempo

for i in range(tempo):
  print(tempo_atual)
  tempo_atual = tempo_atual - 1
  time.sleep(1)
  if tempo_atual == 0:
    print("acabou o tempo")
```

```python
import datetime

agora = datetime.datetime.now()

print(f"Ano atual: {agora.year}")
print(f"Mês atual: {agora.month}")
print(f"Dia de hoje: {agora.day}")
```
