```python
# pedra, papel e tesoura

import random

opcoes = ["pedra", "papel", "tesoura"]
jogador = input("pedra, papel ou tesoura: ")
maquina = random.choice(opcoes)

if jogador == maquina:
  print("empate")

elif jogador == "pedra" and maquina == "tesoura" or\
     jogador == "papel" and maquina == "pedra" or\
     jogador == "tesoura" and maquina == "papel":
     print(f"opção do jogador: {jogador}")
     print(f"opção da maquina: {maquina}")
     print(f"o jogador venceu com {jogador}")

else:
  print(f"opção do jogador: {jogador}")
  print(f"opção da maquina: {maquina}")
  print(f"a maquina venceu com {maquina}")
```

```python
# exercicio 2

frutas = []

quant_frutas = int(input("quantas frutas queres adicionar? "))

for i in range(quant_frutas):
  frutas.append(input("qual fruta quer adicionar? "))

print(f"lista de frutas: \n{frutas}")
```
