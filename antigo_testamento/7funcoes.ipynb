```python
'''
Crie uma funcao chamada "calcular_dano()", ela deve receber duas coisa:
1. o ataque do personagem
2. defesa do inimigo
o dano será o ataque - defesa
se o resultado for menor que 0, o dano deve ser 0
peça os valores utilizando input()
'''

def calcular_dano(ataque, defesa):

  # calculando o dano
  dano = ataque - defesa

  # verificando se o dano é maior ou menor que zero
  if dano <= 0:
    print("você não deu dano, ataque baixo demais!")
  else:
    print(f"você deu {dano} de dano")

calcular_dano(int(input("quanto de ataque você tem? ")),
              int(input("quanto de defesa o inimigo tem? ")))
```
