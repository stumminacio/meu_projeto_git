```python
# Dada a lista (1,3,2,3,4,5,1,5,7,6,8,3,4), crie uma segunda lista apenas com os itens na mesma ordem, mas sem repetição.

lista_1 = [1,3,2,3,4,5,1,5,7,6,8,3,4]
lista_2 = []

for item in lista_1:
  if item not in lista_2:
    lista_2.append(item)

print(f"lista inicial : {lista_1}")
print(f"lista final :   {lista_2}")
```

```python
# DDD
ddd = {
    61: 'Brasilia', 71: 'Salvador', 11: 'São Paulo', 21: 'Rio de Janeiro',
    32: 'Juiz de Fora', 19: 'Campinas', 27: 'Vitoria', 31: 'Belo Horizonte'
}

escolha_ddd = int(input("Qual é seu DDD: "))

if escolha_ddd not in ddd:
  print("Seu DDD não existe!")
else:
  print(ddd[escolha_ddd])
```
