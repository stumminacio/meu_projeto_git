## Trabalhando com imagens

```python
import cv2
from google.colab.patches import cv2_imshow

imagem = cv2.imread("imagem.jpeg")



imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)
imagem_invertida = cv2.bitwise_not(imagem)
imagem_flipada = cv2.flip(imagem, 1)
imagem_redimensionada = cv2.resize(imagem, (960, 516))


cv2_imshow(imagem)
cv2_imshow(imagem_cinza)
cv2_imshow(imagem_invertida)
cv2_imshow(imagem_flipada)
cv2_imshow(imagem_redimensionada)
```
